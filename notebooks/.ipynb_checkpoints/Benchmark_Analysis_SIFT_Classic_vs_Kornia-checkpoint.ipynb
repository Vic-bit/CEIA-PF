```python
# Visualizar Ground Truth (vista superior)
if gt_data:
    x_gt = np.array(gt_data['trajectory']['x'])
    y_gt = np.array(gt_data['trajectory']['y'])
    z_gt = np.array(gt_data['trajectory']['z'])
    times_gt = np.array(gt_data['trajectory']['times'])
    
    # Crear figura
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
    
    # Top-down view
    ax = axes[0]
    ax.plot(x_gt, z_gt, 'b-', linewidth=2, label='Ground Truth')
    ax.plot(x_gt[0], z_gt[0], 'go', markersize=12, label='Start')
    ax.plot(x_gt[-1], z_gt[-1], 'ro', markersize=12, label='End')
    ax.set_xlabel('X (metros)', fontsize=11)
    ax.set_ylabel('Z (metros)', fontsize=11)
    ax.set_title('Ground Truth - Vista Superior (X-Z)', fontweight='bold')
    ax.grid(True, alpha=0.3)
    ax.legend()
    ax.axis('equal')
    
    # Distancia acumulada
    ax = axes[1]
    distances = np.sqrt(np.sum(np.diff(np.array([x_gt, y_gt, z_gt]).T, axis=0)**2, axis=1))
    cumsum_dist = np.concatenate([[0], np.cumsum(distances)])
    ax.plot(times_gt, cumsum_dist, 'b-', linewidth=2)
    ax.set_xlabel('Tiempo (segundos)', fontsize=11)
    ax.set_ylabel('Distancia Acumulada (metros)', fontsize=11)
    ax.set_title('Trayectoria - Distancia vs Tiempo', fontweight='bold')
    ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    print(f"✓ Ground Truth visualizado: {len(x_gt)} frames")
```

```python
# Cargar Ground Truth (poses.txt del dataset KITTI)
import sys
sys.path.insert(0, '../')

# Si ya existe ground_truth_trajectory.json, cargarlo
ground_truth_file = Path('../outputs/benchmarks/ground_truth_trajectory.json')

if ground_truth_file.exists():
    with open(ground_truth_file, 'r') as f:
        gt_data = json.load(f)
    print("✓ Ground Truth cargado desde JSON")
else:
    print("⚠️  ground_truth_trajectory.json no encontrado")
    print("Ejecuta primero: python ground_truth_analysis.py")
    gt_data = None

# Mostrar información
if gt_data:
    print("\n📊 Ground Truth Statistics:")
    stats = gt_data['statistics']
    print(f"  • Frames: {stats['num_frames']}")
    print(f"  • Duración: {stats['duration']:.2f}s")
    print(f"  • Distancia total: {stats['total_distance']:.2f}m")
    print(f"  • FPS: {stats['avg_fps']:.2f}")
```

# Análisis Comparativo: SIFT Classic (NumPy) vs SIFT Kornia (PyTorch)

Este notebook carga y analiza los benchmarks de ambas implementaciones de Visual SLAM.

## 1. Importar Librerías Necesarias

```python
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import numpy as np
from datetime import datetime

# Configurar estilo de visualizaciones
sns.set_style("darkgrid")
plt.rcParams['figure.figsize'] = (12, 6)

print("✓ Librerías importadas exitosamente")
```

## 2. Cargar Archivos JSON

```python
# Definir rutas de los archivos JSON
json_dir = Path('../outputs/benchmarks')
classic_json = json_dir / 'sift_classic.json'
kornia_json = json_dir / 'sift_kornia.json'

# Cargar datos
with open(classic_json, 'r') as f:
    classic_data = json.load(f)
    
with open(kornia_json, 'r') as f:
    kornia_data = json.load(f)

print(f"✓ Datos de SIFT Classic cargados")
print(f"✓ Datos de SIFT Kornia cargados")
print(f"\nArchivos cargados:")
print(f"  - {classic_json}")
print(f"  - {kornia_json}")
```

## 3. Explorar Estructura de Datos

```python
# Mostrar estructura SIFT Classic
print("="*60)
print("SIFT CLASSIC - Estructura de Datos")
print("="*60)
print("\nMetadata:")
for key, value in classic_data['metadata'].items():
    print(f"  {key}: {value}")

print("\nSummary (Resumen):")
for key, value in classic_data['summary'].items():
    print(f"  {key}: {value}")
```

```python
# Mostrar estructura SIFT Kornia
print("="*60)
print("SIFT KORNIA - Estructura de Datos")
print("="*60)
print("\nMetadata:")
for key, value in kornia_data['metadata'].items():
    print(f"  {key}: {value}")

print("\nSummary (Resumen):")
for key, value in kornia_data['summary'].items():
    print(f"  {key}: {value}")
```

## 4. Extraer y Comparar Métricas

```python
# Extraer datos summary de ambas implementaciones
classic_summary = classic_data['summary']
kornia_summary = kornia_data['summary']

# Crear DataFrame para comparación
comparison_df = pd.DataFrame({
    'SIFT Classic': classic_summary,
    'SIFT Kornia': kornia_summary
})

print("\n" + "="*60)
print("COMPARACIÓN DE MÉTRICAS")
print("="*60)
print(comparison_df)
print()
```

## 5. Calcular Diferencias y Ratios

```python
# Calcular diferencias absolutas y porcentuales
print("\n" + "="*60)
print("ANÁLISIS DE DIFERENCIAS")
print("="*60)

metrics = ['total_time', 'avg_fps', 'avg_matches', 'total_ram_mb']
print(f"\n{'Métrica':<20} {'Classic':<15} {'Kornia':<15} {'Diferencia':<15} {'Ratio':<10}")
print("-" * 75)

for metric in metrics:
    classic_val = classic_summary[metric]
    kornia_val = kornia_summary[metric]
    diff = kornia_val - classic_val
    ratio = kornia_val / classic_val if classic_val != 0 else 0
    
    print(f"{metric:<20} {classic_val:<15.2f} {kornia_val:<15.2f} {diff:<15.2f} {ratio:<10.2f}x")
```

## 6. Análisis Detallado por Métrica

```python
# Tiempo Total de Ejecución
print("\n" + "="*60)
print("TIEMPO TOTAL DE EJECUCIÓN")
print("="*60)
classic_time = classic_summary['total_time']
kornia_time = kornia_summary['total_time']

print(f"\nSIFT Classic: {classic_time:.2f} segundos")
print(f"SIFT Kornia:  {kornia_time:.2f} segundos")
print(f"Diferencia:   {kornia_time - classic_time:.2f} segundos")
print(f"Ratio:        {kornia_time / classic_time:.2f}x")

faster_slower = "más lento" if kornia_time > classic_time else "más rápido"
print(f"\n→ SIFT Kornia es {faster_slower} en macOS")
```

```python
# FPS Promedio
print("\n" + "="*60)
print("FPS PROMEDIO (Frames Por Segundo)")
print("="*60)
classic_fps = classic_summary['avg_fps']
kornia_fps = kornia_summary['avg_fps']

print(f"\nSIFT Classic: {classic_fps:.2f} FPS")
print(f"SIFT Kornia:  {kornia_fps:.2f} FPS")
print(f"Diferencia:   {kornia_fps - classic_fps:.2f} FPS")
print(f"Ratio:        {kornia_fps / classic_fps:.2f}x")

faster_slower = "menor" if kornia_fps < classic_fps else "mayor"
print(f"\n→ SIFT Kornia tiene {faster_slower} throughput")
```

```python
# Matches Promedio
print("\n" + "="*60)
print("CORRESPONDENCIAS PROMEDIO (Matches)")
print("="*60)
classic_matches = classic_summary['avg_matches']
kornia_matches = kornia_summary['avg_matches']

print(f"\nSIFT Classic: {classic_matches:.1f} matches/frame")
print(f"SIFT Kornia:  {kornia_matches:.1f} matches/frame")
print(f"Diferencia:   {kornia_matches - classic_matches:.1f} matches")
print(f"Ratio:        {kornia_matches / classic_matches:.2f}x")

better_worse = "mejor" if kornia_matches > classic_matches else "peor"
print(f"\n→ SIFT Kornia tiene {better_worse} calidad de matching")
```

```python
# Consumo de RAM
print("\n" + "="*60)
print("CONSUMO MÁXIMO DE MEMORIA")
print("="*60)
classic_ram = classic_summary['total_ram_mb']
kornia_ram = kornia_summary['total_ram_mb']

print(f"\nSIFT Classic: {classic_ram:.1f} MB")
print(f"SIFT Kornia:  {kornia_ram:.1f} MB")
print(f"Diferencia:   {kornia_ram - classic_ram:.1f} MB")
print(f"Ratio:        {kornia_ram / classic_ram:.2f}x")

more_less = "más" if kornia_ram > classic_ram else "menos"
print(f"\n→ SIFT Kornia usa {more_less} memoria")
```

## 7. Crear Tabla Comparativa Completa

```python
# Crear tabla comparativa con análisis
comparison_analysis = pd.DataFrame({
    'Métrica': ['Tiempo Total (s)', 'FPS Promedio', 'Matches Promedio', 'RAM (MB)'],
    'Classic': [
        f"{classic_summary['total_time']:.2f}",
        f"{classic_summary['avg_fps']:.2f}",
        f"{classic_summary['avg_matches']:.1f}",
        f"{classic_summary['total_ram_mb']:.1f}"
    ],
    'Kornia': [
        f"{kornia_summary['total_time']:.2f}",
        f"{kornia_summary['avg_fps']:.2f}",
        f"{kornia_summary['avg_matches']:.1f}",
        f"{kornia_summary['total_ram_mb']:.1f}"
    ],
    'Ratio (K/C)': [
        f"{kornia_summary['total_time'] / classic_summary['total_time']:.2f}x",
        f"{kornia_summary['avg_fps'] / classic_summary['avg_fps']:.2f}x",
        f"{kornia_summary['avg_matches'] / classic_summary['avg_matches']:.2f}x",
        f"{kornia_summary['total_ram_mb'] / classic_summary['total_ram_mb']:.2f}x"
    ]
})

print("\n" + "="*80)
print("TABLA COMPARATIVA COMPLETA")
print("="*80)
print(comparison_analysis.to_string(index=False))
```

## 8. Visualizar Comparación de Métricas

```python
# Crear visualizaciones
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Comparación SIFT Classic vs SIFT Kornia', fontsize=16, fontweight='bold')

# 1. Tiempo Total
ax1 = axes[0, 0]
implementations = ['Classic', 'Kornia']
times = [classic_summary['total_time'], kornia_summary['total_time']]
colors = ['#1f77b4', '#ff7f0e']
ax1.bar(implementations, times, color=colors, alpha=0.7, edgecolor='black')
ax1.set_ylabel('Tiempo (segundos)', fontsize=11)
ax1.set_title('Tiempo Total de Ejecución', fontweight='bold')
for i, v in enumerate(times):
    ax1.text(i, v + max(times)*0.02, f'{v:.2f}s', ha='center', fontweight='bold')

# 2. FPS Promedio
ax2 = axes[0, 1]
fps_values = [classic_summary['avg_fps'], kornia_summary['avg_fps']]
ax2.bar(implementations, fps_values, color=colors, alpha=0.7, edgecolor='black')
ax2.set_ylabel('FPS', fontsize=11)
ax2.set_title('FPS Promedio', fontweight='bold')
for i, v in enumerate(fps_values):
    ax2.text(i, v + max(fps_values)*0.02, f'{v:.2f}', ha='center', fontweight='bold')

# 3. Matches Promedio
ax3 = axes[1, 0]
matches_values = [classic_summary['avg_matches'], kornia_summary['avg_matches']]
ax3.bar(implementations, matches_values, color=colors, alpha=0.7, edgecolor='black')
ax3.set_ylabel('Matches por Frame', fontsize=11)
ax3.set_title('Correspondencias Promedio', fontweight='bold')
for i, v in enumerate(matches_values):
    ax3.text(i, v + max(matches_values)*0.02, f'{v:.1f}', ha='center', fontweight='bold')

# 4. Consumo de RAM
ax4 = axes[1, 1]
ram_values = [classic_summary['total_ram_mb'], kornia_summary['total_ram_mb']]
ax4.bar(implementations, ram_values, color=colors, alpha=0.7, edgecolor='black')
ax4.set_ylabel('RAM (MB)', fontsize=11)
ax4.set_title('Consumo Máximo de Memoria', fontweight='bold')
for i, v in enumerate(ram_values):
    ax4.text(i, v + max(ram_values)*0.02, f'{v:.1f}MB', ha='center', fontweight='bold')

plt.tight_layout()
plt.savefig('../outputs/benchmarks/comparison_metrics.png', dpi=150, bbox_inches='tight')
plt.show()

print("✓ Gráfica guardada en: outputs/benchmarks/comparison_metrics.png")
```

## 9. Resumen Ejecutivo

```python
print("\n" + "="*80)
print("RESUMEN EJECUTIVO")
print("="*80)

print(f"\n📊 DATASET:")
print(f"   - SIFT Classic: {classic_data['metadata']['total_frames']} frames")
print(f"   - SIFT Kornia: {kornia_data['metadata']['total_frames']} frames")

print(f"\n⏱️ PERFORMANCE:")
print(f"   - SIFT Classic: {classic_summary['total_time']:.2f}s @ {classic_summary['avg_fps']:.2f} FPS")
print(f"   - SIFT Kornia:  {kornia_summary['total_time']:.2f}s @ {kornia_summary['avg_fps']:.2f} FPS")
if kornia_summary['total_time'] > classic_summary['total_time']:
    slowdown_factor = kornia_summary['total_time'] / classic_summary['total_time']
    print(f"   → Kornia es {slowdown_factor:.2f}x más lento (limitación MPS en macOS)")
else:
    speedup_factor = classic_summary['total_time'] / kornia_summary['total_time']
    print(f"   → Kornia es {speedup_factor:.2f}x más rápido")

print(f"\n🎯 ROBUSTEZ (Correspondencias):")
print(f"   - SIFT Classic: {classic_summary['avg_matches']:.1f} matches/frame")
print(f"   - SIFT Kornia:  {kornia_summary['avg_matches']:.1f} matches/frame")
if abs(kornia_summary['avg_matches'] - classic_summary['avg_matches']) < 5:
    print(f"   → Calidad de matching EQUIVALENTE")
elif kornia_summary['avg_matches'] > classic_summary['avg_matches']:
    print(f"   → Kornia tiene MEJOR calidad de matching")
else:
    print(f"   → Classic tiene mejor calidad de matching")

print(f"\n💾 MEMORIA:")
print(f"   - SIFT Classic: {classic_summary['total_ram_mb']:.1f} MB")
print(f"   - SIFT Kornia:  {kornia_summary['total_ram_mb']:.1f} MB")
ram_ratio = kornia_summary['total_ram_mb'] / classic_summary['total_ram_mb']
print(f"   → Ratio: {ram_ratio:.2f}x")

print(f"\n" + "="*80)
print("📝 CONCLUSIONES:")
print("="*80)
print(f"""\n1. ARQUITECTURA:
   - SIFT Classic: NumPy puro (CPU, OpenCV nativo)
   - SIFT Kornia: PyTorch tensorial (GPU/MPS)

2. PERFORMANCE EN MacOS:
   - Kornia es más lento debido a limitaciones de MPS (no soporte completo)
   - En GPU CUDA/NVIDIA sería más rápido
   - La lentitud NO es del método, sino de la aceleración en macOS

3. ROBUSTEZ:
   - Ambas mantienen calidad de tracking comparable
   - Matches similares → Trayectorias/mapas de calidad similar

4. RECOMENDACIÓN ACADÉMICA:
   - SIFT Kornia es viable como alternativa moderna
   - Tensorial architecture → mejor paralelización
   - Portable y extensible (futuro con GPU)
""")
```
